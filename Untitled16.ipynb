What does a single row represent?
A single row represents one invoice line item: A specific product purchased in a given quantity at a given price at a particular time as part of an invoice.

Why do multiple rows share the same Invoice and Customer ID?
Multiple rows may share the same invoice/basket because each invoic can contain multiple different products.
Multiple rows may share the same Customer ID because customers can make multiple purchases over time.

**Possible units of analysis**

**A) Invoice line (row-level)**
At the invoice line level, each row represents a single product purchased within an invoice.

**Gained Information:**
1. Detailed product-level behaviour  
2. Fine-grained view of quantities and prices  

**Lost Information:** 
1. Products within the same invoice are treated as independent  
2. Basket structure and customer intent are ignored  



**B) Invoice / basket level**
At the invoice level, all rows sharing the same Invoice number are aggregated to represent a single purchase event (basket).

**Gained Information:**   
1. Basket size and total invoice value  
2. Transaction-level behaviour  
3. Preserves the transactional structure of the data  

**Lost Information:** 
1. Individual product-level detail  
2. Product mix within a basket  


**C) Customer level**
At the customer level, all invoices associated with a customer are aggregated to represent overall customer behaviour.

**Gained Information:**
1. Customer lifetime value  
2. Purchase frequency and total spend  

**Lost Information:** 
1. Temporal ordering of purchases  
2. Individual transaction-level behaviour  


**Chosen unit of analysis:**

The invoice (basket) level is chosen as the primary unit of analysis for onward work.

This choice preserves the transactional structure of the dataset because it prevents the mistake of treating separate lines from an invoice as separate data entries. It enables researchers to study customer buying behavior through detailed analysis of shopping behavior and shopping cart value and purchase sequence data. Also maintains its ability to generate customer-level features which researchers can use in their subsequent analysis.

```python
from IPython.display import display
import numpy as np

print("Data Quality and Validity Audit")

# 1) Missing or invalid Customer ID

missing_cust = df["Customer ID"].isna().sum()
missing_cust_pct = df["Customer ID"].isna().mean() * 100
invalid_cust = (df["Customer ID"] <= 0).sum()

print("\n1) Missing or invalid Customer ID")
print("Missing Customer IDs:", missing_cust)
print(f"Percentage missing: {missing_cust_pct:.2f}%")
print("Invalid (non-positive) Customer IDs:", invalid_cust)

print("\nSample rows with missing Customer ID:")
display(df[df["Customer ID"].isna()].head())

missing_customer_rows = df[df["Customer ID"].isna()]

print("Rows with missing Customer ID:")
display(missing_customer_rows)

invalid_customer_rows = df[df["Customer ID"] <= 0]

print("Rows with invalid (non-positive) Customer ID:")
display(invalid_customer_rows)

# 2) Cancelled invoices

cancelled_mask = df["Invoice"].astype(str).str.startswith("C")
cancelled_count = cancelled_mask.sum()

print("\n2) Cancelled invoices")
print("Number of cancelled invoice rows:", cancelled_count)

print("\nSample cancelled invoice rows:")
display(df[cancelled_mask].head())

cancelled_invoice_rows = df[df["Invoice"].astype(str).str.startswith("C")]

print("Rows with cancelled invoices:")
display(cancelled_invoice_rows)

# 3) Extreme prices or quantities

print("\n3) Extreme prices or quantities")

print("\nPrice distribution summary:")
display(df["Price"].describe())

print("\nQuantity distribution summary:")
display(df["Quantity"].describe())

# Identify potentially extreme values using simple thresholds
extreme_price_rows = df[df["Price"] > df["Price"].quantile(0.99)]
extreme_quantity_rows = df[np.abs(df["Quantity"]) > df["Quantity"].quantile(0.99)]

print("\nNumber of rows with extreme prices (top 1%):", len(extreme_price_rows))
print("Number of rows with extreme quantities (top 1% by absolute value):", len(extreme_quantity_rows))

print("\nSample rows with extreme prices:")
display(extreme_price_rows.head())

print("\nSample rows with extreme quantities:")
display(extreme_quantity_rows.head())

price_threshold = df["Price"].quantile(0.99)

extreme_price_rows = df[df["Price"] > price_threshold]

print(f"Rows with extreme prices (Price > {price_threshold:.2f}):")
display(extreme_price_rows)

quantity_threshold = df["Quantity"].abs().quantile(0.99)

extreme_quantity_rows = df[df["Quantity"].abs() > quantity_threshold]

print(f"Rows with extreme quantities (|Quantity| > {quantity_threshold}):")
display(extreme_quantity_rows)
```

**Issue 1) Missing or invalid Customer ID:**

**Description:**
A substantial number of rows contain missing values in the Customer ID column. Additionally, a small number of rows contain invalid (non-positive) customer identifiers.

**Why it matters:**
In order to establish connections between customer transactions and their corresponding individual customers, Customer IDs is required. The absence of correct identifiers makes it impossible to perform customer-level analysis because it blocks the ability to track purchase habits and customer value and behavioral changes throughout time.

**Planned handling:**
The analysis will eliminate all entries which have either missing or incorrect Customer IDs because they need customer-specific assessment. The system maintains these records for two main purposes which include transactional and invoice-level data analysis without needing customer identification information.

**Issue 2) Cancelled invoices:**

**Description:**
The transaction cancellation status of some invoices becomes evident through their "C" prefix identifiers. The dataset contains two types of purchase data which include both missing purchase information and full purchase details.

**Why it matters:**
Because it shows cancelled invoices as inactive documents which do not indicate any finished business transactions. The evaluation of customer purchasing patterns through financial performance data would generate wrong results because these businesses should stay away from such research activities. The evaluation of customer purchasing patterns and revenue data would generate wrong results because these organizations need to stay out of such research studies.

**Planned handling:**
The analysis of completed purchases and revenue generation will exclude all rows which correspond to cancelled invoices.

**Issue 3) Extreme prices or quantities:**

Description:
The distributions of Price and Quantity contain extreme values relative to the majority of observations.

**Why it matters:**
Extreme values can disproportionately influence aggregate measures such as total revenue or average basket size. However, such values may also represent legitimate bulk purchases or high-value items rather than data errors.

**Planned handling:**
Extreme prices and quantities will be flagged for further inspection rather than automatically removed. Their impact will be assessed before deciding whether to retain, exclude, or treat them separately in subsequent analyses.



```python

print(" Task 4: Minimal, Justified Cleaning ")

# Decision 1: Parse dates correctly

print("\n[Decision 1] Parse InvoiceDate to datetime")

print("Before parsing:")
print("InvoiceDate dtype:", df["InvoiceDate"].dtype)
print("Sample InvoiceDate values:")
display(df["InvoiceDate"].head(5))

df_clean = df.copy()
df_clean["InvoiceDate"] = pd.to_datetime(df_clean["InvoiceDate"], errors="coerce")

print("\nAfter parsing:")
print("InvoiceDate dtype:", df_clean["InvoiceDate"].dtype)
print("Invalid/unparseable dates (NaT):", df_clean["InvoiceDate"].isna().sum())
print("Date range (min -> max):", df_clean["InvoiceDate"].min(), "->", df_clean["InvoiceDate"].max())

print("\nSample parsed InvoiceDate values:")
display(df_clean[["InvoiceDate"]].head(5))

# Decision 2: Separate returns from purchases

print("\n[Decision 2] Separate returns (Quantity < 0) from purchases (Quantity > 0)")

print("Before separation:")
print("Total rows:", len(df_clean))
print("Negative quantity rows (returns):", (df_clean["Quantity"] < 0).sum())
print("Positive quantity rows (purchases):", (df_clean["Quantity"] > 0).sum())
print("Zero quantity rows:", (df_clean["Quantity"] == 0).sum())

purchases = df_clean[df_clean["Quantity"] > 0].copy()
returns = df_clean[df_clean["Quantity"] < 0].copy()
zero_qty = df_clean[df_clean["Quantity"] == 0].copy()

print("\nAfter separation:")
print("Purchase rows:", len(purchases))
print("Return rows:", len(returns))
print("Zero-quantity rows (kept separate for now):", len(zero_qty))

print("\nSample purchase rows:")
display(purchases.head())

print("\nSample return rows:")
display(returns.head())
```

**4) Minimal, Justified Cleaning**

**Cleaning decision 1:** Parse InvoiceDate correctly

**Justification:**  
InvoiceDate must be a proper datetime type to support temporal EDA (e.g., revenue over time) and any time-based aggregation.

**Evidence:** 
Before cleaning, InvoiceDate is stored as a non-datetime type. After conversion using `pd.to_datetime`, the column becomes datetime and any invalid/unparseable entries are identified as `NaT`.

---

**Cleaning decision 2:** Separate returns from purchases

**Justification:**  
Negative quantities represent returns, not purchases. Mixing them with purchases would distort analyses such as revenue trends and basket metrics.

**Evidence:** 
Counts of negative and positive quantities are shown before separation, and the dataset is split into purchases and returns, with example rows displayed from each.


```python
import matplotlib.pyplot as plt

print(" Task 5: Exploratory Data Analysis ")

df_eda = purchases.copy()

df_eda["Revenue"] = df_eda["Quantity"] * df_eda["Price"]

# EDA 1: Temporal pattern — revenue over time

daily_revenue = (
    df_eda
    .groupby(df_eda["InvoiceDate"].dt.date)["Revenue"]
    .sum()
)

plt.figure(figsize=(10,4))
daily_revenue.plot()
plt.title("Daily Revenue Over Time")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.show()
```

**Question:** How does total revenue vary over time?  
**Key pattern:** Revenue fluctuates over time with visible peaks and troughs, suggesting seasonality and changes in purchasing activity.  
**Limitation:** This analysis does not control for the number of active customers or promotions during each period.

```python
# EDA 2: Transactional pattern — basket size (invoice-level)

basket_size = (
    df_eda
    .groupby("Invoice")["Quantity"]
    .sum()
)
display(basket_size.describe())
```

### EDA 2: Basket size distribution (invoice-level)
**Question:** How many items are typically purchased per invoice?  
**Key pattern:** Most invoices contain a relatively small number of items, while a smaller number of invoices have very large basket sizes.  
**Limitation:** Basket size alone does not capture product diversity or item value.


```python
# EDA 3: Transactional pattern — invoice value

invoice_value = (
    df_eda
    .groupby("Invoice")["Revenue"]
    .sum()
)
plt.figure(figsize=(6,4))
invoice_value.hist(bins=50)
plt.title("Distribution of Invoice Values")
plt.xlabel("Invoice Value")
plt.ylabel("Frequency")
plt.show()
```

**Question:** How much do customers typically spend per transaction?  
**Key pattern:** Invoice values are right-skewed, with many low-value transactions and a small number of high-value invoices.  
**Limitation:** Extreme values may reflect bulk purchases or unusual transactions that require further investigation.


```python
# EDA 4: Country pattern — revenue by country

country_revenue = (
    df_eda
    .groupby("Country")["Revenue"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)
display(country_revenue)
```

**Question:** How is revenue distributed across countries?  
**Key pattern:** Revenue is heavily concentrated in a small number of countries, indicating strong geographic imbalance.  
**Limitation:** Results are influenced by dataset coverage and do not reflect market size or population differences.


```python
# EDA 5: Product pattern — most frequently sold products

top_products = (
    df_eda
    .groupby("Description")["Quantity"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)
display(top_products)
```

**Question:** Which products are sold in the highest quantities?  
**Key pattern:** A small number of products account for a large proportion of total units sold.  
**Limitation:** High sales volume does not necessarily imply high revenue or profitability.

**Task 6) Reflection and planning**

**Three insights from Week 1 analysis:**

1. Transactional structure matters
The dataset contains information at the invoice line level because each transaction (invoice) includes multiple rows in the dataset. The analysis of each row as separate transactions would produce incorrect results about customer shopping patterns because it would create artificial transaction numbers.
2. Data quality issues are non-trivial.
The data contains missing customer identifiers together with cancelled invoices and returns that appear as negative quantities and abnormal extreme values. The identification and proper management of these problems becomes essential because they create major problems for revenue calculation and customer-level analysis.
3. Strong skewness and imbalance exist
Both basket size and invoice value distributions are highly right-skewed, and revenue is heavily concentrated in a small number of countries. The results show that using averages as the only metric fails to provide complete information about customer and transaction patterns.

**Two assumptions or risks:**

1. Assumption of valid extreme values
Extreme prices and quantities are assumed to potentially represent legitimate bulk or high-value purchases. The presence of data errors in these values will produce incorrect statistical results when researchers perform their calculations and build their models.
2. Risk from missing Customer IDs
A large number of transactions lack customer identifiers. Any customer-level analysis implicitly assumes that the remaining identifiable customers are representative, which may not be the case.

**Chosen unit of analysis for onward work:**

The invoice (basket) level is chosen as the primary unit of analysis for onward work.
This preserves the transactional structure of the data and allows meaningful analysis of basket size, invoice value, and purchasing behaviour without conflating individual line items with complete transactions.

**Brief note on the modelling task for later stages:**

The cleaned invoice-level data becomes available for use in customer segmentation and demand analysis and revenue prediction tasks during subsequent stages. The modeling process should use clustering methods to group customer data by shopping patterns and time-series forecasting techniques which analyze sales patterns throughout different time periods.
